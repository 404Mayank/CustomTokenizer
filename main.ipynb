```python
%pip install kagglehub

from pathlib import Path
import kagglehub

# Download the dataset
dataset_path = kagglehub.dataset_download(
    "madhavendraoo7/book-text-files"
)

print("Dataset downloaded to:", dataset_path)

dataset_path = Path(dataset_path)

for file_path in dataset_path.rglob("*"):
    print(file_path)
```

```python
books = []

for file_path in sorted(dataset_path.glob("*.txt")):
    text = file_path.read_text(
        encoding="utf-8",
        errors="replace"
    )
    books.append(text)

for i, book in enumerate(books):
    print(f"Book {i+1} :\n")
    print(book[:100] + "....")
```

```python
books_inUTF8 = []

for i, book in enumerate(books):
    books_inUTF8.append(book.encode("utf-8"))
    print(books_inUTF8[i][:100])
```

```python
def occuranceCounter(input):
    counts = {}
    for pair in zip(input[:], input[1:]):
        counts[pair] = counts.get(pair, 0) + 1
    return counts

counts = []

for i, book in enumerate(books_inUTF8):
    counts.append(occuranceCounter(book))
    print(f"Book {i+1} :")
    print(sorted( ((v,k) for (k,v) in counts[i].items()) , reverse=True))
```

```python
max_counts = []
for count in counts:
    max_counts.append(max(count, key=count.get))
for max_count in max_counts:
    print(max_count)
```

```python
def mergeTokens(tokens, tokenPair, replaceWith):
    output = []
    i = 0
    length = len(tokens)
    while i < length:
        if i < (length - 1) and tokens[i] == tokenPair[0] and tokens[i + 1] == tokenPair[1]:
            output.append(replaceWith)
            i = i + 2
        else:
            output.append(tokens[i])
            i = i + 1
    return output

# print(mergeTokens([1,2,3,3,4,5,6,7,4,3,3,5], (3,3), 69))

books_inUTF8_merged = []

for i, book in enumerate(books_inUTF8):
    books_inUTF8_merged.append(mergeTokens(book, max_counts[i], 256))
    print(len(book), len(books_inUTF8_merged[i]))
```

```python
text = []
for book in books_inUTF8:
    text.extend(list(book))
len(text)
```

```python
def initVocab():
    return { i: bytes([i]) for i in range(256) }

initVocab()
```

```python
vocab_size = 500
i = 0

def getVocab(vocab_size, tokens):
    vocabStart = 256
    currTokens = tokens
    vocab = initVocab()
    merges = {}
    if (len(vocab) < vocab_size):
        for i in range(vocab_size - len(vocab)):
            counts = occuranceCounter(currTokens)
            if not counts:
                break
            topPair = max(counts, key=counts.get)
            if (counts[topPair] <= 1):
                break
            currTokens = mergeTokens(currTokens, topPair, (vocabStart + i))
            merges[topPair] = (vocabStart + i)
            vocab[vocabStart + i] = vocab[topPair[0]] + vocab[topPair[1]]
            print(f"merged {topPair} with {counts[topPair]} ocr. | len: {len(currTokens)}")
    else:
        print("vocab already maxxed!")
    return vocab, merges, currTokens
        

vocab, merges, final = getVocab(vocab_size, text)
```

```python
def encode(text):
    tokens =  text.encode("utf-8")
    print(f"Iniital size: {len(tokens)} | ", end='')
    while(True):
        counts = occuranceCounter(tokens)
        commonPairs = [k for k in merges if k in counts]
        if not commonPairs:
            break
        tokens = mergeTokens(tokens, commonPairs[0], merges[commonPairs[0]])
    print(f"Final Size: {len(tokens)}")
    print(f"Compression Ratio: {'%.3f'%(float(len(text)/len(tokens)))}")
    return tokens
def decode(tokens):
    tokenUnmerged = b"".join(vocab[token] for token in tokens)
    return tokenUnmerged.decode("utf-8", errors='replace')
decode(encode(books[0])) == books[0]
```
